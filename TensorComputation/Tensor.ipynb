On a device, tensor is interpreted as a datatype, meaning every tensor, is an object of Tensor class. A tensor, same as a numpy array or a python tuple or list, can compropmise any datatype, including other tensors, other objects of different datatypes, etc. Any object or instance of a class (a datatype) indicates the region of the device storage space, i.e. registers inside which being numerically stored (the concept of self in python). Tensors represent multidimensional arrays, althogh they only are objects stored as bits inside the memory.

For a tensor with indices
\begin{gather*}
  \big(i_0, i_1, \ldots, i_{n-1}\big)
\end{gather*}
and strides
\begin{gather*}
  \big(s_0, s_1, \ldots, s_{n-1}\big)
\end{gather*}
the storage position is
\begin{gather*}
  \textit{offset} = \textit{storage offset} + \sum_{k=0}^{n-1} i_ks_k
\end{gather*}

```python
tensor1 = torch.tensor([[1.0, 9.0], [23.0, 14.0]], requires_grad=True)
tensor1_transposed = tensor1.permute(1,0)

tensor2 = torch.tensor([[1.0,9.0], [23.0,14.0], [31.0,19.0]], requires_grad=True)
tensor2_transposed = tensor2.T

print(tensor1_transposed.shape)
print(tensor2_transposed.shape)
```

```python
def random_shuffle_transpose(shape, n=1, seed=None):
    if n < 0:
        raise ValueError("n must be non-negative")

    generator = (
        torch.Generator().manual_seed(seed)
        if seed is not None
        else None
    )

    tensor = torch.rand(shape, generator=generator)

    if tensor.ndim < 2:
        raise ValueError("shape must have at least two dimensions")

    original_shape = tensor.shape

    for _ in range(n):
        indices = torch.randperm(tensor.numel(), generator=generator)
        tensor = tensor.flatten()[indices].reshape(original_shape)

    tensor = tensor.transpose(0, 1) if tensor.ndim == 2 else tensor.mT

    return tensor
```
