**Notation.** Vectors are bold and column-oriented: $\boldsymbol{x}=(x_1,\dots,x_n)^{\mathsf T}\in\mathbb{R}^n$. Throughout, $f:\mathbb{R}^n\to\mathbb{R}$ is scalar-valued and $\boldsymbol{f}:\mathbb{R}^n\to\mathbb{R}^m$ is vector-valued.

---

## 1. The Univariate Derivative and the Differential

$$
y = f(x)
\quad\Longrightarrow\quad
\frac{dy}{dx} = \frac{df(x)}{dx} = f'(x)
\qquad\Longleftrightarrow\qquad
dy = f'(x)\,dx .
$$

The derivative $f'(x)$ is the local proportionality constant between an infinitesimal input perturbation $dx$ and the induced output perturbation $dy$. Equivalently, the **differential** $dy = f'(x)\,dx$ expresses the best *linear* approximation of $f$ at $x$:

$$
f(x+dx) = f(x) + f'(x)\,dx + o(|dx|).
$$

This reading, the derivative as a *linear map* rather than a number, is the one that generalizes to higher dimensions.

---

## 2. Scalar-Valued Functions of Several Variables: the Gradient

For $y = f(\boldsymbol{x}) = f(x_1,x_2,\dots,x_n)$, the derivative is the $1\times n$ **row vector** of partial derivatives, and the **gradient** is its transpose, a column vector:

$$
\boldsymbol{\nabla}_{\boldsymbol{x}} f(\boldsymbol{x})
= \left(\frac{\partial f(\boldsymbol{x})}{\partial \boldsymbol{x}}\right)^{\mathsf T}
= \begin{pmatrix}
\dfrac{\partial f(\boldsymbol{x})}{\partial x_1}\\[8pt]
\dfrac{\partial f(\boldsymbol{x})}{\partial x_2}\\[4pt]
\vdots\\[4pt]
\dfrac{\partial f(\boldsymbol{x})}{\partial x_n}
\end{pmatrix}
\in \mathbb{R}^{n}.
$$

The **total differential** aggregates the contribution of each coordinate perturbation:

$$
df = \sum_{i=1}^{n}\frac{\partial f}{\partial x_i}\,dx_i
= \big\langle \boldsymbol{\nabla}_{\boldsymbol{x}} f,\; d\boldsymbol{x} \big\rangle
= \big(\boldsymbol{\nabla}_{\boldsymbol{x}} f\big)^{\mathsf T} d\boldsymbol{x}.
$$

**Interpretation.** The gradient is the Riesz representer of the derivative: it converts the linear functional $d\boldsymbol{x}\mapsto df$ into an inner product. By the Cauchy–Schwarz inequality, $df$ is maximal when $d\boldsymbol{x}\parallel\boldsymbol{\nabla}_{\boldsymbol{x}}f$. Hence the gradient points in the direction of steepest ascent, which underlies gradient-based optimization.

---

## 3. Vector-Valued Functions: the Jacobian Matrix

Let $\boldsymbol{y} = \boldsymbol{f}(\boldsymbol{x})$ with $m$ component functions:

$$
\boldsymbol{y}
=
\begin{pmatrix} y_1\\ y_2\\ \vdots\\ y_m \end{pmatrix}
=
\begin{pmatrix}
f_1(x_1,\dots,x_n)\\
f_2(x_1,\dots,x_n)\\
\vdots\\
f_m(x_1,\dots,x_n)
\end{pmatrix}.
$$

Its derivative is the $m\times n$ **Jacobian matrix**

$$
\boldsymbol{J}
= \frac{\partial \boldsymbol{y}}{\partial \boldsymbol{x}}
= \frac{\partial \boldsymbol{f}(\boldsymbol{x})}{\partial \boldsymbol{x}}
=
\begin{pmatrix}
\dfrac{\partial f_1}{\partial x_1} & \dfrac{\partial f_1}{\partial x_2} & \cdots & \dfrac{\partial f_1}{\partial x_n}\\[10pt]
\dfrac{\partial f_2}{\partial x_1} & \dfrac{\partial f_2}{\partial x_2} & \cdots & \dfrac{\partial f_2}{\partial x_n}\\[6pt]
\vdots & \vdots & \ddots & \vdots\\[6pt]
\dfrac{\partial f_m}{\partial x_1} & \dfrac{\partial f_m}{\partial x_2} & \cdots & \dfrac{\partial f_m}{\partial x_n}
\end{pmatrix},
\qquad
J_{ij} = \left(\frac{\partial \boldsymbol{f}}{\partial \boldsymbol{x}}\right)_{ij} = \frac{\partial f_i}{\partial x_j}.
$$

Alternative notations for the same object are $\boldsymbol{J}^{\,\boldsymbol{y}}_{\boldsymbol{x}}$ and $\boldsymbol{J}_{\boldsymbol{f}}(\boldsymbol{x})$. The convention is that **rows index outputs** and **columns index inputs**. This is the *numerator layout*, in which $\boldsymbol{J}\in\mathbb{R}^{m\times n}$.

---

## 4. Row-wise Structure and the Differential of a Vector Function

Each row of $\boldsymbol{J}$ is the derivative (a transposed gradient) of one component function:

$$
\boldsymbol{J}
=
\begin{pmatrix}
(\boldsymbol{\nabla}_{\boldsymbol{x}} f_1)^{\mathsf T}\\[4pt]
(\boldsymbol{\nabla}_{\boldsymbol{x}} f_2)^{\mathsf T}\\[4pt]
\vdots\\[4pt]
(\boldsymbol{\nabla}_{\boldsymbol{x}} f_m)^{\mathsf T}
\end{pmatrix}
=
\begin{pmatrix}
\dfrac{\partial f_1(\boldsymbol{x})}{\partial \boldsymbol{x}}\\[8pt]
\dfrac{\partial f_2(\boldsymbol{x})}{\partial \boldsymbol{x}}\\[4pt]
\vdots\\[4pt]
\dfrac{\partial f_m(\boldsymbol{x})}{\partial \boldsymbol{x}}
\end{pmatrix}.
$$

The **differential** of $\boldsymbol{f}$ is the linear map $d\boldsymbol{x}\mapsto \boldsymbol{J}\,d\boldsymbol{x}$:

$$
d\boldsymbol{y} = \boldsymbol{J}\,d\boldsymbol{x}
\quad\Longleftrightarrow\quad
\begin{pmatrix} dy_1\\ dy_2\\ \vdots\\ dy_m \end{pmatrix}
=
\begin{pmatrix}
\dfrac{\partial f_1}{\partial x_1} & \cdots & \dfrac{\partial f_1}{\partial x_n}\\[8pt]
\dfrac{\partial f_2}{\partial x_1} & \cdots & \dfrac{\partial f_2}{\partial x_n}\\[4pt]
\vdots & \ddots & \vdots\\[4pt]
\dfrac{\partial f_m}{\partial x_1} & \cdots & \dfrac{\partial f_m}{\partial x_n}
\end{pmatrix}
\begin{pmatrix} dx_1\\ dx_2\\ \vdots\\ dx_n \end{pmatrix}.
$$

Evaluating the matrix–vector product row by row shows that each output perturbation is an inner product of a gradient with the input perturbation:

$$
d\boldsymbol{y}
=
\begin{pmatrix}
(\boldsymbol{\nabla}_{\boldsymbol{x}} f_1)^{\mathsf T} d\boldsymbol{x}\\[4pt]
(\boldsymbol{\nabla}_{\boldsymbol{x}} f_2)^{\mathsf T} d\boldsymbol{x}\\[4pt]
\vdots\\[4pt]
(\boldsymbol{\nabla}_{\boldsymbol{x}} f_m)^{\mathsf T} d\boldsymbol{x}
\end{pmatrix}
=
\begin{pmatrix}
\langle \boldsymbol{\nabla}_{\boldsymbol{x}} f_1,\, d\boldsymbol{x}\rangle\\[4pt]
\langle \boldsymbol{\nabla}_{\boldsymbol{x}} f_2,\, d\boldsymbol{x}\rangle\\[4pt]
\vdots\\[4pt]
\langle \boldsymbol{\nabla}_{\boldsymbol{x}} f_m,\, d\boldsymbol{x}\rangle
\end{pmatrix},
\qquad
dy_i = \sum_{j=1}^{n}\frac{\partial f_i}{\partial x_j}\,dx_j .
$$

## 5. A Neural Network as a Composite Multivariable Function

A neural network is a highly structured, high-dimensional multivariable function $\boldsymbol{f}:\mathbb{R}^{d_0}\to\mathbb{R}^{d_n}$. It maps a large input vector to an output vector through a composition of simple parametrised layers.

### 5.1 A single computational layer

Define the **pre-activation** $\boldsymbol{u}^{(\ell)}$ and the **activation** $\boldsymbol{a}^{(\ell)}$:

$$
\boldsymbol{u}^{(\ell)} = \mathbf{W}^{(\ell)}\boldsymbol{a}^{(\ell-1)}+\boldsymbol{b}^{(\ell)},
\qquad
\boldsymbol{a}^{(\ell)} = f^{(\ell)}\!\big(\boldsymbol{u}^{(\ell)}\big)
= f^{(\ell)}\Big(\mathbf{W}^{(\ell)}\boldsymbol{a}^{(\ell-1)}+\boldsymbol{b}^{(\ell)}\Big).
$$

The affine map can be absorbed into one linear map by **bias augmentation**:

$$
\boldsymbol{a}^{(\ell)}
= f^{(\ell)}\Big(\mathbf{A}^{(\ell)}\,\boldsymbol{a}^{\prime(\ell-1)}\Big),
\qquad
\mathbf{A}^{(\ell)}=\big[\,\mathbf{W}^{(\ell)}\;\big|\;\boldsymbol{b}^{(\ell)}\,\big],
\qquad
\boldsymbol{a}^{\prime(\ell-1)}=\begin{pmatrix}\boldsymbol{a}^{(\ell-1)}\\ 1\end{pmatrix}.
$$

### 5.2 A feedforward network of depth $n$

Composing $n$ such layers gives the **feedforward neural network (FFNN)**:

$$
\underbrace{\boldsymbol{y} = \boldsymbol{a}^{(n)} = f^{(n)}\Big(\mathbf{W}^{(n)}\Big(f^{(n-1)}\Big(\mathbf{W}^{(n-1)}\Big(f^{(n-2)}\Big(\mathbf{W}^{(n-2)}\Big(\cdots \Big(f^{(1)} \Big(\mathbf{W}^{(1)} \boldsymbol{x}+\boldsymbol{b}^{(1)} \Big)\Big) \cdots\Big)+\boldsymbol{b}^{(n-2)}\Big)\Big)+\boldsymbol{b}^{(n-1)}\Big)\Big)+\boldsymbol{b}^{(n)}\Big)}_{\text{An FFNN}}
$$

Equivalently, in compositional form,

$$
\boldsymbol{f}=\boldsymbol{f}_n\circ\boldsymbol{f}_{n-1}\circ\cdots\circ\boldsymbol{f}_1,
\qquad
\boldsymbol{f}_\ell(\boldsymbol{a})=f^{(\ell)}\!\big(\mathbf{W}^{(\ell)}\boldsymbol{a}+\boldsymbol{b}^{(\ell)}\big).
$$

---

## 6. The Jacobian of the Network

The **Jacobian** measures how the activation of a computational unit changes when the activation of a prior unit changes. Entry $(i,j)$ of $\partial\boldsymbol{a}^{(\ell)}/\partial\boldsymbol{a}^{(\ell-1)}$ is the sensitivity of unit $i$ in layer $\ell$ to unit $j$ in layer $\ell-1$.

### 6.1 Chain rule factorisation

By the chain rule, the end-to-end Jacobian $\mathbf{J}=\partial\boldsymbol{a}^{(n)}/\partial\boldsymbol{x}\in\mathbb{R}^{d_n\times d_0}$ factorises into a product of layer-to-layer Jacobians:

$$
\mathbf{J}
= \frac{\partial \boldsymbol{a}^{(n)}}{\partial \boldsymbol{x}}
= \frac{\partial \boldsymbol{a}^{(n)}}{\partial \boldsymbol{a}^{(n-1)}}\,
  \frac{\partial \boldsymbol{a}^{(n-1)}}{\partial \boldsymbol{a}^{(n-2)}}\cdots
  \frac{\partial \boldsymbol{a}^{(1)}}{\partial \boldsymbol{x}} .
$$

Each layer-to-layer Jacobian is itself a product of two simpler Jacobians, by routing through the pre-activation $\boldsymbol{u}^{(\ell)}$:

$$
\frac{\partial \boldsymbol{a}^{(\ell)}}{\partial \boldsymbol{a}^{(\ell-1)}}
= \frac{\partial \boldsymbol{a}^{(\ell)}}{\partial \boldsymbol{u}^{(\ell)}}\,
  \frac{\partial \boldsymbol{u}^{(\ell)}}{\partial \boldsymbol{a}^{(\ell-1)}}
= \mathbf{D}^{(\ell)}\,\mathbf{W}^{(\ell)},
$$

where

$$
\mathbf{D}^{(\ell)}=\frac{\partial \boldsymbol{a}^{(\ell)}}{\partial \boldsymbol{u}^{(\ell)}}=\operatorname{diag}\!\Big(f^{(\ell)\prime}\big(\boldsymbol{u}^{(\ell)}\big)\Big),
\qquad
\frac{\partial \boldsymbol{u}^{(\ell)}}{\partial \boldsymbol{a}^{(\ell-1)}}=\mathbf{W}^{(\ell)}.
$$

$\mathbf{D}^{(\ell)}$ is **diagonal** because $f^{(\ell)}$ acts elementwise, so $a_i^{(\ell)}$ depends only on $u_i^{(\ell)}$. (This does not hold for non-elementwise activations such as softmax.)

Substituting, and expanding the first chain rule through every pre-activation, gives

$$
\mathbf{J}
= \frac{\partial \boldsymbol{a}^{(n)}}{\partial \boldsymbol{x}}
= \frac{\partial \boldsymbol{a}^{(n)}}{\partial \boldsymbol{u}^{(n)}}\,
  \frac{\partial \boldsymbol{u}^{(n)}}{\partial \boldsymbol{a}^{(n-1)}}\cdots
  \frac{\partial \boldsymbol{a}^{(1)}}{\partial \boldsymbol{u}^{(1)}}\,
  \frac{\partial \boldsymbol{u}^{(1)}}{\partial \boldsymbol{x}}
= \mathbf{D}^{(n)}\mathbf{W}^{(n)}\cdots\mathbf{D}^{(1)}\mathbf{W}^{(1)}
= \prod_{\ell=n}^{1}\mathbf{D}^{(\ell)}\mathbf{W}^{(\ell)} .
$$

The product is **ordered**: matrix multiplication is non-commutative, so the index runs from $\ell=n$ on the left to $\ell=1$ on the right.

### 6.2 The transposed Jacobian

Transposing reverses the order of the factors:

$$
\mathbf{J}^{\mathsf T}
= \left(\frac{\partial \boldsymbol{a}^{(1)}}{\partial \boldsymbol{x}}\right)^{\mathsf T}\cdots
  \left(\frac{\partial \boldsymbol{a}^{(n-1)}}{\partial \boldsymbol{a}^{(n-2)}}\right)^{\mathsf T}
  \left(\frac{\partial \boldsymbol{a}^{(n)}}{\partial \boldsymbol{a}^{(n-1)}}\right)^{\mathsf T}
= \prod_{\ell=1}^{n}\mathbf{W}^{(\ell)\mathsf T}\mathbf{D}^{(\ell)\mathsf T}
= \prod_{\ell=1}^{n}\mathbf{W}^{(\ell)\mathsf T}\mathbf{D}^{(\ell)} .
$$

The last equality uses $\mathbf{D}^{(\ell)\mathsf T}=\mathbf{D}^{(\ell)}$, since diagonal matrices are symmetric.

**Interpretation.** Applying $\mathbf{J}$ to a vector propagates a perturbation **forward** from input to output (forward-mode AD, a Jacobian–vector product). Applying $\mathbf{J}^{\mathsf T}$ to a vector propagates a sensitivity **backward** from output to input (reverse-mode AD, a vector–Jacobian product). Backpropagation is the repeated application of $\mathbf{W}^{(\ell)\mathsf T}\mathbf{D}^{(\ell)}$, from the last layer to the first.

## 7. Derivatives with Respect to the Parameters

The Jacobian above differentiates with respect to the **input** $\boldsymbol{x}$. Training requires derivatives with respect to the **parameters** $\mathbf{W}^{(\ell)}$ and $\boldsymbol{b}^{(\ell)}$. The same two-step routing through $\boldsymbol{u}^{(\ell)}$ applies.

### 7.1 First layer

Take the first layer,

$$
\boldsymbol{a}^{(1)} = f^{(1)}\!\left(\mathbf{W}^{(1)}\boldsymbol{x}+\boldsymbol{b}^{(1)}\right) = f^{(1)}\!\left(\boldsymbol{u}^{(1)}\right).
$$

Its derivatives with respect to the input, the weights, and the bias are

$$
\frac{\partial \boldsymbol{a}^{(1)}}{\partial \boldsymbol{x}}
= \frac{\partial \boldsymbol{a}^{(1)}}{\partial \boldsymbol{u}^{(1)}}\,\frac{\partial \boldsymbol{u}^{(1)}}{\partial \boldsymbol{x}},
\qquad
\frac{\partial \boldsymbol{a}^{(1)}}{\partial \mathbf{W}^{(1)}}
= \frac{\partial \boldsymbol{a}^{(1)}}{\partial \boldsymbol{u}^{(1)}}\,\frac{\partial \boldsymbol{u}^{(1)}}{\partial \mathbf{W}^{(1)}},
\qquad
\frac{\partial \boldsymbol{a}^{(1)}}{\partial \boldsymbol{b}^{(1)}}
= \frac{\partial \boldsymbol{a}^{(1)}}{\partial \boldsymbol{u}^{(1)}}\,\frac{\partial \boldsymbol{u}^{(1)}}{\partial \boldsymbol{b}^{(1)}} .
$$

Since $\boldsymbol{u}^{(1)}$ is affine in its arguments, the inner factors are elementary:

$$
\frac{\partial \boldsymbol{u}^{(1)}}{\partial \boldsymbol{x}}=\mathbf{W}^{(1)},
\qquad
\frac{\partial \boldsymbol{u}^{(1)}}{\partial \boldsymbol{b}^{(1)}}=\mathbf{I},
\qquad
\frac{\partial u^{(1)}_i}{\partial W^{(1)}_{jk}}=\delta_{ij}\,x_k .
$$

Hence

$$
\frac{\partial \boldsymbol{a}^{(1)}}{\partial \boldsymbol{x}}=\mathbf{D}^{(1)}\mathbf{W}^{(1)},
\qquad
\frac{\partial \boldsymbol{a}^{(1)}}{\partial \boldsymbol{b}^{(1)}}=\mathbf{D}^{(1)} .
$$

The derivative $\partial\boldsymbol{u}^{(1)}/\partial\mathbf{W}^{(1)}$ is a **third-order tensor** (vector with respect to matrix). It is sparse, because $u_i$ depends only on row $i$ of $\mathbf{W}$, and in practice it is never formed explicitly.

### 7.2 A deep layer: the layer-by-layer recursion

Consider the stack of layers up to depth 11 as an example:

$$
\left.
\begin{aligned}
  \boldsymbol{a}^{(11)} &= f^{(11)}\!\left(\mathbf{W}^{(11)}\boldsymbol{a}^{(10)}+\boldsymbol{b}^{(11)}\right) = f^{(11)}\!\left(\boldsymbol{u}^{(11)}\right)\\
  &\;\;\vdots  \\
  \boldsymbol{a}^{(2)} &= f^{(2)}\!\left(\mathbf{W}^{(2)}\boldsymbol{a}^{(1)}+\boldsymbol{b}^{(2)}\right) = f^{(2)}\!\left(\boldsymbol{u}^{(2)}\right)\\
  \boldsymbol{a}^{(1)} &= f^{(1)}\!\left(\mathbf{W}^{(1)}\boldsymbol{x}+\boldsymbol{b}^{(1)}\right) = f^{(1)}\!\left(\boldsymbol{u}^{(1)}\right)
\end{aligned}
\right\}
$$

The weights $\mathbf{W}^{(1)}$ influence $\boldsymbol{a}^{(11)}$ only **indirectly**, through every intermediate layer. The chain rule therefore threads the sensitivity through the whole stack:

$$
\frac{\partial \boldsymbol{a}^{(11)}}{\partial \mathbf{W}^{(1)}}
= \frac{\partial \boldsymbol{a}^{(11)}}{\partial \boldsymbol{u}^{(11)}}\,
  \frac{\partial \boldsymbol{u}^{(11)}}{\partial \boldsymbol{a}^{(10)}}\,
  \frac{\partial \boldsymbol{a}^{(10)}}{\partial \boldsymbol{u}^{(10)}}\,
  \frac{\partial \boldsymbol{u}^{(10)}}{\partial \boldsymbol{a}^{(9)}}
  \cdots
  \frac{\partial \boldsymbol{a}^{(2)}}{\partial \boldsymbol{u}^{(2)}}\,
  \frac{\partial \boldsymbol{u}^{(2)}}{\partial \boldsymbol{a}^{(1)}}\,
  \frac{\partial \boldsymbol{a}^{(1)}}{\partial \boldsymbol{u}^{(1)}}\,
  \frac{\partial \boldsymbol{u}^{(1)}}{\partial \mathbf{W}^{(1)}} .
$$

Substituting $\mathbf{D}$ and $\mathbf{W}$ for each local Jacobian collapses this to

$$
\frac{\partial \boldsymbol{a}^{(11)}}{\partial \mathbf{W}^{(1)}}
= \underbrace{\mathbf{D}^{(11)}\mathbf{W}^{(11)}\,\mathbf{D}^{(10)}\mathbf{W}^{(10)}\cdots\mathbf{D}^{(2)}\mathbf{W}^{(2)}}_{\displaystyle\prod_{k=11}^{2}\mathbf{D}^{(k)}\mathbf{W}^{(k)}}\;
  \mathbf{D}^{(1)}\,\frac{\partial \boldsymbol{u}^{(1)}}{\partial \mathbf{W}^{(1)}} .
$$

### 7.3 General form

For an arbitrary pair of layers $\ell\le L$, the sensitivity of the layer-$L$ activation to the layer-$\ell$ weights is

$$
\frac{\partial \boldsymbol{a}^{(L)}}{\partial \mathbf{W}^{(\ell)}}
= \Bigg(\prod_{k=L}^{\ell+1}\mathbf{D}^{(k)}\mathbf{W}^{(k)}\Bigg)\,
  \mathbf{D}^{(\ell)}\,
  \frac{\partial \boldsymbol{u}^{(\ell)}}{\partial \mathbf{W}^{(\ell)}},
\qquad
\frac{\partial \boldsymbol{a}^{(L)}}{\partial \boldsymbol{b}^{(\ell)}}
= \Bigg(\prod_{k=L}^{\ell+1}\mathbf{D}^{(k)}\mathbf{W}^{(k)}\Bigg)\,\mathbf{D}^{(\ell)} .
$$

The first factor is the Jacobian of the sub-network from layer $\ell$ to layer $L$, the same object as in Section 6, truncated.
